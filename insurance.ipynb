```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sc
```

```python
df = pd.read_csv("Healthcare.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['Age'] = df['Age'].fillna(25)
```

```python
df['Age'] = df['Age'].replace("45 years",45)
df['Age'] = df['Age'].replace("fifty",45)
df['Age'] = df['Age'].replace("unknown",26)
```

```python
df['Age'] = df['Age'].replace(26,47)
```

```python
df['Age'] = df['Age'].astype(int)
```

```python
df.head(2)
```

```python
df['Blood_Pressure_mmHg'] = df['Blood_Pressure_mmHg'].fillna(78)
```

```python
df.head()
```

```python
df['Blood_Pressure_mmHg'] = df['Blood_Pressure_mmHg'].replace('unknown',110)
```

```python
df['Blood_Pressure_mmHg'] = df['Blood_Pressure_mmHg'].replace('130 mmHg',115)
```

```python
df['Blood_Pressure_mmHg'] = df['Blood_Pressure_mmHg'].replace('140/90',120)
```

```python
df['Blood_Pressure_mmHg'] = df['Blood_Pressure_mmHg'].astype(int)
```

```python
df['Gender'].unique()
```

```python
df['Gender'] = df['Gender'].replace(" female","Female")
df['Gender'] = df['Gender'].replace("female ","Female")
df['Gender'] = df['Gender'].replace("FEMALE","Female")
df['Gender'] = df['Gender'].replace(" male","Male")
df['Gender'] = df['Gender'].replace("Male ","Male")
```

```python
df['Gender'].isnull().sum()
```

```python
df.dropna(subset=['Gender'], inplace=True)
```

```python
df.head()
```

```python
df['BMI'] = df['BMI'].fillna(17)
```

```python
df['BMI'] = df['BMI'].replace("unknown",18)
df['BMI'] = df['BMI'].replace("thirty",30)
df['BMI'] = df['BMI'].replace("28.5 kg/m2",29)
```

```python
df['BMI'] = df['BMI'].astype(float)
```

```python
df['Disease_Status'].unique()
```

```python
df['Disease_Status'] = df['Disease_Status'].replace(" HEART DISEASE","Heart Disease")
df['Disease_Status'] = df['Disease_Status'].replace("diabetes + heart disease","Diabetes + Heart Disease")
df['Disease_Status'] = df['Disease_Status'].replace("diabetes","Diabetes")
df['Disease_Status'] = df['Disease_Status'].replace("diabetes","Diabetes")
df['Disease_Status'] = df['Disease_Status'].replace("Healthy ","Healthy")
```

```python
df.dropna(subset=['Disease_Status'], inplace=True)
```

```python
df['Disease_Status'].isnull().sum()
```

```python
df.shape
```

```python
df = df.to_csv("Clean_healthcare.csv")
```
